### **1. Trasformare cicli in list comprehension**

Fai tre esempi di cicli producono una lista e poi riscrivi lo stesso codice usando la list comprehension

```python
#PRIMA LISTA
quadrati = []

for n in range(1,5):
    quadrati.append(n**2)

#con list comprehension
quadrati=[n**2 for n in range(1,5)]


#SECONDA LISTA
pari = []

for n in range(1, 10):
    if n%2 == 0:
        pari.append(n)

#con list comprehension
pari = [n for n in range(1,5) if n%2 == 0]


#TERZA LISTA
dispari = []

for n in range(1,10):
    if n%2 != 0:
        dispari.append(n)

#con list comprehension
dispari = [n for n in range(1, 10) if n%2 != 0]
```
