# 5. Analisi di un dataset di video giochi
1. Quanti videogiochi sono stati pubblicati
2. Quali sono i generi più popolari? Mostralo con un bar plot
3. Plotta l'evoluzione del numero di giochi pubblicati nel tempo
4. Prova a riprodurre questo plot 




```python
# Import dei pacchetti
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

"""Caricamento del dataset
url = 'https://zenodo.org/record/5898311/files/vgsales.csv'NON ME LO APRE 
df = pd.read_csv(url)
df.head()
"""
#alternativa:
url = 'https://raw.githubusercontent.com/CodingwithO/VideoGameSales/master/vgsales.csv'
df = pd.read_csv(url)
df.head()
```

```python
#videogiochi pubblicati
df.shape[0] #righe nel dataset

len(df['Name'].unique())

print('Righe nel dataset:', df.shape[0])
print('Titoli diversi:', len(df['Name'].unique()))
```

```python
#generi popolari con barplot
sns.countplot(x='Genre', data=df)
plt.title('Numero di giochi pubblicati per genere')
plt.xticks(rotation=45)
plt.show()
```

il grafico mostra che i generi più popolari sono ACTION, SPORTS, MISC, ROLE-PLAYING e SHOOTER

```python
# evoluzione numero di giochi per anno

giochi_anno = df.groupby('Year').size() #raggruppo per anno e conto le righe

giochi_anno.plot(kind = 'line')
plt.title('Giochi pubblicati per anno')
plt.xlabel('Anno')
plt.ylabel('Numero di giochi')
plt.show()
```

```python
#riproduzione del plot
regioni = ['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales'] #seleziono solo le regioni presenti nel plot

top5 = df['Genre'].value_counts().head(5).index #noto che i generi presenti nel plot sono quelli più popolari
top5

#Sommo le vendite per genere, tengo i 5 generi con più giochi, e li rimetto in ordine alfabetico.
vendite = df.groupby('Genre')[regioni].sum().loc[top5].sort_index()
vendite

vendite.plot(kind='bar', stacked=True)
plt.title('Vendite per regione per genere')
plt.ylabel('Milioni di unità')
plt.show()
```
